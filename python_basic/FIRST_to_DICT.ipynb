```python
n = int(input("How many Fibonacci numbers? "))

a, b = 1, 1
count = 0

if n <= 0:
    print("Please enter a positive integer.")
elif n == 1:
    print(f"Fibonacci sequence: {a}")
else:
    print("Fibonacci sequence:")
    while count < n:
        print(a, end=" ")
        a, b = b, a + b
        count += 1
```
